```python
import os

# Crear la carpeta .streamlit si no existe
os.makedirs(".streamlit", exist_ok=True)

# Crear un archivo config.toml para desactivar los webscokets y solucionar el bloqueo en blanco
config_content = """
[server]
enableCORS = false
enableXsrfProtection = false
headless = true
address = "0.0.0.0"
port = 8501

[browser]
gatherUsageStats = false
serverAddress = "localhost"

[theme]
primaryColor = "#0A365C"
# backgroundColor = "#F4F6F9"
secondaryBackgroundColor = "#EAECEF"
textColor = "#0F172A"
font = "sans serif"
"""

with open(".streamlit/config.toml", "w") as f:
    f.write(config_content.strip())

print("✅ Configuración de Streamlit actualizada para evitar bloqueos de red.")
```

```python
import time
import os
from IPython.display import IFrame

# La app se lanza siempre desde la carpeta del proyecto (clon de Git o copia antigua)
for carpeta in ("/workspace/SIMD-Lab/equipo12", "/workspace/equipo12"):
    if os.path.exists(os.path.join(carpeta, "frontend", "app.py")):
        os.chdir(carpeta)
        break

# ==============================================================================
# CONFIGURACIÓN DE DATOS (Configura aquí: "test_ui" o "real")
# ==============================================================================
os.environ["AUDIT_DATA_MODE"] = "real"  # Opciones: "test_ui" o "real"
# ==============================================================================

# 2. Buscar automáticamente la ruta del python del .venv en las carpetas cercanas
venv_python = os.path.abspath(".venv/bin/python") if os.path.exists(".venv/bin/python") else None
for root, dirs, files in ([] if venv_python else os.walk(".")):
    if "python" in files and ".venv" in root:
        venv_python = os.path.abspath(os.path.join(root, "python"))
        break

if not venv_python or not os.path.exists(venv_python):
    if os.path.exists("../.venv/bin/python"):
        venv_python = os.path.abspath("../.venv/bin/python")

if not venv_python or not os.path.exists(venv_python):
    print("❌ No se ha podido localizar automáticamente el binario de .venv. ¿En qué carpeta lo creaste?")
else:
    print(f"✅ ¡Entorno .venv localizado con éxito en: {venv_python}")
    print(f"⚙️ Modo de datos configurado: {os.environ.get('AUDIT_DATA_MODE')}")
    
    # 3. Lanzar Streamlit pasando explícitamente la variable de entorno
    cmd = (
        f"AUDIT_DATA_MODE={os.environ['AUDIT_DATA_MODE']} PYTHONPATH=. "
        f"{venv_python} -m streamlit run frontend/app.py "
        f"--server.port=8501 --server.address=0.0.0.0 --server.headless=true &"
    )
    get_ipython().system_raw(cmd)
    
    time.sleep(3)
    print("🚀 ¡Servidor Streamlit arrancado utilizando tu entorno virtual!")
```
